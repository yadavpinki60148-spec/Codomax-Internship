```python
# Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Data
subjects = ["Maths", "Chemistry", "BEEE", "English", "Graphics"]
marks = [85, 78, 82, 90, 75]

# Bar Chart
plt.bar(subjects, marks)
plt.title("Marks of Students")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()

# Line Chart
plt.plot(subjects, marks, marker="o")
plt.title("Marks Trend")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()

# Pie Chart
plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Marks Distribution")
plt.show()

# Histogram
plt.hist(marks, bins=5)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()

# Scatter Plot
study_hours = [2, 3, 4, 5, 6]
test_marks = [55, 65, 70, 82, 90]

plt.scatter(study_hours, test_marks)
plt.title("Study Hours and Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()

# Seaborn Bar Chart
sns.barplot(x=subjects, y=marks)
plt.title("Subject Wise Marks")
plt.show()
```
