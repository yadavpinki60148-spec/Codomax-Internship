```python
# Python & Data Science Fundamentals


# 1. Variables and Data Types
name = "Student"
age = 20
marks = 85.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Marks:", marks)
print("Student:", is_student)

print(type(name))
print(type(age))
print(type(marks))
print(type(is_student))


# 2. Conditional Statements and Loops
marks = 85

if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)

for i in range(1, 6):
    print(i)


# 3. Functions
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

marks_list = [80, 75, 90, 85, 70]
average = calculate_average(marks_list)

print("Average marks:", average)


# 4. Lists and Dictionaries
subjects = ["Python", "Maths", "English", "Data Science"]

print("Subjects:", subjects)
print("First subject:", subjects[0])

student = {
    "name": "Student",
    "age": 20,
    "course": "Data Science",
    "score": 85
}

print("Student Details:")
for key, value in student.items():
    print(key, ":", value)


# 5. File Handling
file_name = "student_notes.txt"

with open(file_name, "w") as file:
    file.write("Python and Data Science Fundamentals\n")
    file.write("This file was created using Python file handling.\n")

with open(file_name, "r") as file:
    content = file.read()

print(content)


# 6. Basic Data Science Example
student_scores = {
    "A": 78,
    "B": 92,
    "C": 65,
    "D": 88,
    "E": 74
}

scores = list(student_scores.values())

print("Average Score:", sum(scores) / len(scores))
print("Highest Score:", max(scores))
print("Lowest Score:", min(scores))

print("Students scoring 80 or above:")

for student, score in student_scores.items():
    if score >= 80:
        print(student, score)
```
