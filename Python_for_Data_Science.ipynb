```python
import pandas as pd
import numpy as np

# Creating a small student dataset

data = {
    "Name": ["Asha", "Rahul", "Neha", "Aman", "Priya", "Ravi", "Sneha", "Vikas"],
    "Age": [20, 21, np.nan, 22, 20, 23, 21, np.nan],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "CSE", "ECE", "IT"],
    "Marks": [85, 78, 92, np.nan, 88, 67, 95, 74],
    "Attendance": [90, 82, 95, 76, np.nan, 80, 98, 85]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)


# Saving the data as a CSV file

df.to_csv("student_data.csv", index=False)

print("\nCSV file saved successfully.")


# Reading the CSV file

data = pd.read_csv("student_data.csv")

print("\nFirst few records:")
print(data.head())


# Checking the data

print("\nNumber of rows and columns:")
print(data.shape)

print("\nColumn names:")
print(data.columns)

print("\nBasic information:")
data.info()


# Checking missing values

print("\nMissing values:")
print(data.isnull().sum())


# Filling missing values

data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Marks"] = data["Marks"].fillna(data["Marks"].mean())
data["Attendance"] = data["Attendance"].fillna(data["Attendance"].mean())

print("\nData after filling missing values:")
print(data)


# Basic analysis

print("\nAverage marks:", round(data["Marks"].mean(), 2))
print("Highest marks:", data["Marks"].max())
print("Lowest marks:", data["Marks"].min())

print("\nAverage attendance:", round(data["Attendance"].mean(), 2))


# Department wise marks

print("\nAverage marks by department:")
print(data.groupby("Department")["Marks"].mean().round(2))


# Students having marks above 80

print("\nStudents with marks above 80:")
print(data[data["Marks"] > 80])


# Sorting students according to marks

print("\nStudents sorted by marks:")
print(data.sort_values("Marks", ascending=False))


print("\nModule 2 completed.")
```
