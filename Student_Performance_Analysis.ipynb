```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Study_Hours": [2, 3, 4, 5, 6, 2, 7, 8, 4, 6],
    "Marks": [45, 50, 58, 65, 72, 48, 80, 88, 60, 75],
    "Attendance": [70, 75, 80, 85, 90, 72, 95, 96, 82, 91]
}

df = pd.DataFrame(data)
df
```

```python
print(df.shape)
print(df.describe())
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study_Hours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
print("Key Insights:")
print("1. Students who study more generally score higher marks.")
print("2. Higher attendance is generally linked with better marks.")
print("3. Student H has the highest marks.")
```
